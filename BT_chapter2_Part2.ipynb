```python
van_ban = input("Nhập thông điệp cần mã hóa: ")

ma_hoa = ""

for ky_tu in van_ban:
    # Nếu là chữ hoa (A-Z)
    if 'A' <= ky_tu <= 'Z':
        ma_hoa += chr((ord(ky_tu) - ord('A') + 3) % 26 + ord('A'))
    # Nếu là chữ thường (a-z)
    elif 'a' <= ky_tu <= 'z':
        ma_hoa += chr((ord(ky_tu) - ord('a') + 3) % 26 + ord('a'))
    # Nếu không phải chữ cái → giữ nguyên
    else:
        ma_hoa += ky_tu

print("Thông điệp sau khi mã hóa:", ma_hoa)
```

```python
def ma_hoa_caesar(van_ban, dich):
    ket_qua = ""
    for ky_tu in van_ban:
        if 'A' <= ky_tu <= 'Z':  # Chữ hoa
            ket_qua += chr((ord(ky_tu) - ord('A') + dich) % 26 + ord('A'))
        elif 'a' <= ky_tu <= 'z':  # Chữ thường
            ket_qua += chr((ord(ky_tu) - ord('a') + dich) % 26 + ord('a'))
        else:  # Giữ nguyên ký tự khác
            ket_qua += ky_tu
    return ket_qua

def giai_ma_caesar(van_ban, dich):
    return ma_hoa_caesar(van_ban, -dich)  # Dịch ngược lại

def main():
    lua_chon = input("Bạn muốn (M)ã hóa hay (G)iải mã? ").strip().lower()
    van_ban = input("Nhập thông điệp: ")
    dich = int(input("Nhập số ký tự dịch chuyển: "))

    if lua_chon == 'm':
        ket_qua = ma_hoa_caesar(van_ban, dich)
        print("Thông điệp sau khi mã hóa: ", ket_qua)
    elif lua_chon == 'g':
        ket_qua = giai_ma_caesar(van_ban, dich)
        print("Thông điệp sau khi giải mã: ", ket_qua)
    else:
        print("Lựa chọn không hợp lệ! Vui lòng nhập 'M' hoặc 'G'.")

if __name__ == "__main__":
    main()
```

```python
def cac_danh_sach_con(lst):
    ket_qua = []
    n = len(lst)

    for i in range(n + 1):
        for j in range(i + 1, n + 1):
            ket_qua.append(lst[i:j])

    ket_qua.insert(0, [])
    return ket_qua

ds = [1, 2, 3]
print("Các danh sách con của", ds, "là:")
print(cac_danh_sach_con(ds))
```

```python
def dinh_dang_danh_sach(tu_list):
    n = len(tu_list)
    if n == 0:
        return ""
    elif n == 1:
        return tu_list[0]
    elif n == 2:
        return f"{tu_list[0]} and {tu_list[1]}"
    else:
        return ", ".join(tu_list[:-1]) + " and " + tu_list[-1]

print("Nhập các từ (nhấn Enter để dừng):")
ds_tu = []
while True:
    tu = input("> ").strip()
    if tu == "":
        break
    ds_tu.append(tu)
    
ket_qua = dinh_dang_danh_sach(ds_tu)
print("\nChuỗi sau khi định dạng:")
print(ket_qua)
```
